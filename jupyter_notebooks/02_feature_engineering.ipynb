# Feature Engineering

## Objectives

- Create new features that may improve model performance.
- Transform variables where required.
- Prepare the dataset for machine learning.

## Inputs

- Cleaned dataset from the Data Cleaning notebook.

## Outputs

- Engineered features.
- Feature-engineered dataset for modelling.

```python
import pandas as pd

# Load cleaned dataset from Notebook 1
df = pd.read_csv("cleaned_train.csv")

# Drop rows with missing target
df = df.dropna(subset=["SalePrice"])

df.head()
```

## Feature Engineering

Feature engineering is used to create new variables
that may improve predictive performance.

```python
# Total square footage

df["TotalSF"] = (
    df["TotalBsmtSF"]
    + df["1stFlrSF"]
    + df["2ndFlrSF"]
)

# Total porch area

df["TotalPorchSF"] = (
    df["OpenPorchSF"]
    + df["EnclosedPorch"]
    + df["3SsnPorch"]
    + df["ScreenPorch"]
)

# Property age at sale

df["AgeAtSale"] = (
    df["YrSold"]
    - df["YearBuilt"]
)

# Total bathrooms

df["TotalBath"] = (
    df["FullBath"]
    + (0.5 * df["HalfBath"])
    + df["BsmtFullBath"]
    + (0.5 * df["BsmtHalfBath"])
)

df.head()
```

## Verify Engineered Features

Inspect the newly created features to confirm that feature engineering
was applied successfully and the new variables contain plausible values.

```python
df[
    [
        "TotalSF",
        "TotalPorchSF",
        "AgeAtSale",
        "TotalBath"
    ]
].head()
```

```python
# Save engineered dataset

df.to_csv(
    "engineered_train.csv",
    index=False
)

print("engineered_train.csv saved successfully.")
```
